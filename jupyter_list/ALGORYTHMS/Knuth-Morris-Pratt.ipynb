```python
def compute_lps_array(pattern):
    """
    Computes the Longest Proper Prefix which is also Suffix (LPS) array for the given pattern.
    The LPS array stores, for each prefix of the pattern, the length of its longest proper prefix
    that is also a suffix of that prefix.
    """
    m = len(pattern)
    lps = [0] * m  # Initialize LPS array with zeros
    length = 0  # Length of the previous longest prefix suffix
    i = 1

    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1
    return lps

def kmp_search(text, pattern):
    """
    Performs KMP search to find all occurrences of the pattern in the text.
    Returns a list of starting indices where the pattern is found.
    """
    n = len(text)
    m = len(pattern)

    if m == 0:  # Empty pattern always matches at all positions
        return list(range(n + 1))
    if n == 0:  # Empty text cannot contain a non-empty pattern
        return []

    lps = compute_lps_array(pattern)
    
    i = 0  # Pointer for text
    j = 0  # Pointer for pattern
    occurrences = []

    while i < n:
        if pattern[j] == text[i]:
            i += 1
            j += 1

        if j == m:
            occurrences.append(i - j)  # Pattern found at this index
            j = lps[j - 1]  # Move j to the next possible match using LPS array
        elif i < n and pattern[j] != text[i]:
            if j != 0:
                j = lps[j - 1]  # Mismatch, use LPS array to shift pattern
            else:
                i += 1  # No prefix to match, move to the next character in text
    return occurrences

# Example Usage:
text = "ABABDABACDABABCABAB"
pattern = "ABABCABAB"
found_indices = kmp_search(text, pattern)
print(f"Pattern found at indices: {found_indices}")

text2 = "AAAAA"
pattern2 = "AA"
found_indices2 = kmp_search(text2, pattern2)
print(f"Pattern found at indices: {found_indices2}")

text3 = "ABCDEFG"
pattern3 = "XYZ"
found_indices3 = kmp_search(text3, pattern3)
print(f"Pattern found at indices: {found_indices3}")
```
