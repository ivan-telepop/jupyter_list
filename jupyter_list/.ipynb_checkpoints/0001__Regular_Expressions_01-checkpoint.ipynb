```python
# 1. Символы и квантификаторы
# Паттерн	Значение	Пример
# .	любой символ (кроме \n)	a.b → acb, arb
# ^	начало строки	^Hello — ищет строки, начинающиеся с Hello
# $	конец строки	world$ — ищет строки, оканчивающиеся на world
# *	0 или более повторов	ab* → a, ab, abb
# +	1 или более повторов	ab+ → ab, abb
# ?	0 или 1 повтор	colou?r → color, colour
# {n}	ровно n повторов	\d{4} → 2025
# {n,}	минимум n повторов	\d{2,} → 12, 12345
# {n,m}	от n до m повторов	\d{2,4} → 12, 1234

# 2. Множества символов
# Паттерн	Значение
# [abc]	любой символ из списка
# [^abc]	любой символ, кроме перечисленных
# [a-z]	любой символ от a до z
# [A-Za-z0-9_]	буквы и цифры
# \d	цифра (0–9)
# \D	не цифра
# \w	буква, цифра или _
# \W	всё, кроме \w
# \s	пробел, табуляция, перенос
# \S	всё, кроме пробельных символов

# 3. Группировка и альтернативы
# Паттерн	Значение
# (abc)	группа (запоминает совпадение)
# (?:abc)	группа без запоминания
# `a	b`
# (?P<name>...)	именованная группа
# \1, \2	ссылка на предыдущие группы

# 4. Ленивая (lazy) и жадная (greedy) выборка
# Паттерн	Значение
# .*	жадный — захватывает всё
# .*?	ленивый — минимальный захват
# .+?	ленивый, 1+ символов

# 5. Спецконструкции
# Паттерн	Значение
# (?=...)	позитивный lookahead (перед чем-то)
# (?!...)	негативный lookahead (не перед чем-то)
# (?<=...)	позитивный lookbehind (после чего-то)
# (?<!...)	негативный lookbehind
# (?#комментарий)	комментарий
# (?i)	флаг: регистронезависимый поиск
# (?m)	флаг: многострочный режим
# (?s)	флаг: . включает \n

```

```python
str1 = 'ABCD'
str2 = '1234'

wrd = 'Alpho Betha Control Doo'
de = 't'

tbl = wrd.maketrans(str1,str2,de)

wrd.translate(tbl)
```

```python
# Специальные символы,
# '.' - любой символ;
# '^' - началу строки;
# '$' - конец строки;
# '*' - 0 или более повторений;
# '+' - 1 или более повторений;
# '?' - 0 или 1 повторений;
# '*?', '+?','??' - ограничение жадности;
# '*+', '++', '?+' - притяжательные квантификаторы, (новое в Python 3.11);
# '{m}' - m повторений;
# '{m,n}' - как можно больше повторений в промежутке от m до n ;
# '{m,n}?' - как можно меньше повторений в промежутке от m до n;
# '{m,n}+' - притяжательная версия квантификатора выше, (новое в Python 3.11);
# '\' - экранирование специальных символов;
# '[]' - символьный класс;
# '|' - или;
# '(...)' - группа с захватом;
```

```python
# Флаги, используемые в функции re.compile()
# Функция search() модуля re
# Функция match() модуля re
# Функция fullmatch() модуля re
# Функция finditer() модуля re
# Функция split() модуля re
# Функция findall() модуля re
# Функция sub() модуля re
# Функция subn() модуля re
# Функция escape() модуля re
# Функция purge() модуля re
# Исключение error() модуля re
# Объект регулярного выражения Pattern модуля re
# Объект совпадения с шаблоном Match модуля re
```

```python
import re

text = "My email is test@mail.com and phone is +123-456-789."

# Email
emails = re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", text)
print(emails)  # ['test@mail.com']

# Телефон
phones = re.findall(r"\+\d{3}-\d{3}-\d{3}", text)
print(phones)  # ['+123-456-789']

# Найти слово "email" в начале строки (игнор регистра)
if re.match(r"(?i)^my", text):
    print("Начинается с 'my'")
```

```python
import re

phone_pattern = r"^\+?\d{1,3}?[-.\s]?\(?\d{3}\)?[-.\s]?\d{3}[-.\s]?\d{4}$"

# Assumes at least a first and last name with a space in between
name_pattern = r"^[a-zA-Z]"


body = "This text has few names also has phone numbers that are users with names Anna Kovaleva, Kostya Makoedov, Vanya Goncharov, so they has fone numbers like +79855203082, +7 365-667-22-55, 89885441382"

res = re.match(name_pattern,body)

res
```

```python
# {n,m}	от n до m повторов	\d{2,4} → 12, 1234

text = 'The test text is +987656789 1 22 33 4444 then Ivan +TTT--xx 00909'


for i in range(1,9):
    res = re.findall(f'\d[0-{i}]',text)
    print(res)

res
```
