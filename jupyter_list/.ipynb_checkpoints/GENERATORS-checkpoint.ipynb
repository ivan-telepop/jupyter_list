```python
# Interesting pattern ROUND-ROBIN - how to make sequence or carousel making next()


def gen(items):
    for i in items:
        yield i
                

#### ----------------
g1 = gen([1,2,3,4,5])
g2 = gen(['one','two','three','four','five'])
g3 = gen(['один','два','три','четыре','пять'])

tasks = [g1,g2,g3]
# while cycle
while tasks:
    current = tasks.pop(0)
    try:
        x = next(current)
        print(x)
        tasks.append(current)
    except StopIteration:
        pass
```

```python
obj = {
    'key1': {'one1': 1,'two1': 2,'three1': 3,'four1': 4,'five1': 5}, 
       'key2': {'one2': 1,'two2': 2,'three2': 3,'four2': 4,'five2': 5},
       'key3': {'one3': 1,'two3': 2,'three3': 3,'four3': 4,'five3': 5},
       'key4': {'one4': 1,'two4': 2,'three4': 3,'four4': 4,'five4': 5},
       'key5': {'one5': 1,'two5': 2,'three5': 3,'four5': 4,'five5': 5}
      }

def gener(items):
    for one in items.items():
        for i in one:
            yield i

gg = gener(obj)




print(next(gg))
```

```python
# So cool pattern !!!

# Мы создаем как бы генераторный контейнер куда можно вбрасывать любые объекты
# На сколько фантазия позволит исключения и обработки ситуаций 


def bar():
    while True:
        res = yield
        if res == "X":
            raise BaseException("Shit Happened Exception")
        print('Got something',res)


foo = bar()



next(foo)


foo.send('HolaHola')
foo.send('XXX')

foo.send('X')
```

```python
# If we will set next to yield return we will see such behaviour

def bar(x):
    if x <= 20:
        yield x
    return "Hello world"

foo = bar(30)


next(foo)

# next(foo)
    
        
```

```python
# Delegation in generators
def chain_gen(a,b):
    yield from a
    yield from b


a = [1,2,3,4,5]
b = [6,7,8,9,0]
x = [ i for i in range(10,30) ]




foo = chain_gen([a,b], chain_gen(x,['qqq','www','eee','rrr']))

print(*foo)
```
