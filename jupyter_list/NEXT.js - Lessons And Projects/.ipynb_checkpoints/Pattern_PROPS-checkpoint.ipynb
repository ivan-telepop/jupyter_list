```typescript
// Допустим у нас лежит компонент `myapp/app/components/square-component.tsx` который что то выводит
// Мы в него прокидываем пропсы через аргументы - name,age,phone,nums
// а далее  через пропсы протягиваем нужное 

export default function SqwareComponent({name,age,phone,nums}){
    return(
        ...
            <p>{name}</p>
            <p>{age}</p>
            <p>{phone}</p>
            <div>
        // пример с рендерингом списка map()
            <ul>
                {
                nums.map((value,index) => {
                return<li key={index}>{index} -- {value}</li>
                
            })}

            </ul>
        // пример с рендерингом списка через map()
            </div>
            
        </main>
        </div>
    )
    
}


// Далее используем компонент - допустим что у нас переиспользуемый компонент используется в модуле `app/page.tsx` 
// мы просто прокидываем в пропсы нужное
// как в примере ниже:

        <SqwareComponent name="Ivan" age={46} phone={'+7985524363'} nums={['one','two','three','four','five','six']}/>
```
