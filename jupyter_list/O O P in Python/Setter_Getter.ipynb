```python
# Getters (Accessors):
# Methods used to retrieve the value of an attribute. 
# They often begin with get_ or simply use the attribute name.


class Tester1:
    def __init__(self,attr1,attr2):
        self.attr1 = attr1
        self.attr2 = attr2
    def attr2(self):
        "Getter method as an method which retrieves attr"
        return self.attr2


t1 = Tester1(100,200)

t1.attr2
```

```python
# Setters (Mutators):
# Methods used to set or modify the value of an attribute. 
# They often begin with set_ or use the attribute name. 



class Tester2:
    def __init__(self,one,two):
        self.one = one
        self.two = two
    def set_one(self,value):
        self.one = value

foo = Tester2(22,55)

print('Initial value of one is:',foo.one)

foo.set_one(55555)

print('Renewed value of one is:', foo.one)
```
