```python
# ARGS - любой объект оборачивает в tuple

# KWARGS - именованные параметры, или просто словарь


def greet_me(*args,**kwargs):
    print("Type of args:", type(args))
    print("Type of kwargs:", type(kwargs))
    for key, value in kwargs.items():
        print(f"{key} = {value}")
    print(args)

greet_me([1,2,3],name="yasoob", age=30, city="Oslo")
        
```
