```python
# Protected modifying

class Protected:
    """ Protected example of the class"""
    _key = None
    _secret = None
    _salt = None
    def __init__(self,key,secret,salt):
        """Init method that initiates instance"""
        self._key = key
        self._secret = secret
        self._salt = salt
    def _make_it_hash(self):
        """Protected method that returns the hashed key arg"""
        return hash(self._key)


prObj = Protected(key="Mykey",secret='123',salt='the salt')

print("__dict__ are filled with params:",prObj.__dict__)

# The way to retrieve value from the arg
prObj._key

# The method usage 
prObj._make_it_hash()
```

```python
# Private mangling

class Private:
    """ Private example of the class"""
    __key = None
    __secret = None
    __salt = None
    def __init__(self,key,secret,salt):
        """Init method that initiates instance"""
        self.__key = key
        self.__secret = secret
        self.__salt = salt
    def __make_it_hash(self):
        """Private method that returns the hashed key arg"""
        return hash(self.__key)


privO = Private(key='123',secret='super secret',salt='salty')

print("Class Private has attrs:", privO._Private__key, privO._Private__secret,privO._Private__salt)
```

```python
# MRO protected

class ParentA:
    """ Private Parent example of the class"""
    __key1 = None
    __secret1 = None
    __salt1 = None
    def __init__(self,key,secret,salt):
        """Init method that initiates instance"""
        self.__key1 = key
        self.__secret1 = secret
        self.__salt1 = salt
    def __make_it_hash1(self):
        """Private method that returns the hashed key arg"""
        return hash(self.__key1)

class ParentB:
    """ Private Parent example of the class"""
    __key2 = None
    __secret2 = None
    __salt2 = None
    def __init__(self,key,secret,salt):
        """Init method that initiates instance"""
        self.__key2 = key
        self.__secret2 = secret
        self.__salt2 = salt
    def __make_it_hash2(self):
        """Private method that returns the hashed key arg"""
        return hash(self.__key2)



# --- Child class with private attrs 


class Child(ParentA, ParentB):
    def __init__(self,key,secret,salt):
        self.key = key
        self.secret = secret
        self.salt = salt
        super().__init__(key,secret,salt)
        


ch = Child('a','b','c')


# print("Child class has attrs:",ch.__dict__)

print("Attr ParentB:",ch._ParentB__make_it_hash2())


print("Attr ParentA:",ch._ParentA__make_it_hash1())






        
```
