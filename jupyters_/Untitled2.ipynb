```python
import math

# первые 100 цифр числа pi (включая 3)
pi_digits_str = "31415926535897932384626433832795028841971693993751058209749445923078164062862089986280348253421170679"

def pi_square_challenge(digits):
    selected = pi_digits_str[:digits] 
    sum_of_squares = sum(int(d)**2 for d in selected)
    return math.ceil(math.sqrt(sum_of_squares))

# Пример:
print(pi_square_challenge(5))  # Вывод: 8
```
