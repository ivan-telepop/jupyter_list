```python
# PORT SCAN EXAMPLE
import socket

ports = [22,80,443,445,3389,80,8000,5367,8080,6379,5432,1433,3306]

def return_lan(addr):
    return f"192.168.0.{addr}"

addr_interval = range(1,256)

def scan_ports(host):
    for port in ports:
        sock = socket.socket()
        sock.settimeout(0.1)
        if sock.connect_ex((host, port))==1:
            print(f"Detected port: {port} on {host}")
            with open('opened-ports.txt', 'a',encoding='utf-8') as file:
                file.write(f"Detected port: {port} on {host} \n")
        sock.close()

for addr in addr_interval:
    cur_addr = return_lan(addr)
    print(cur_addr)
    scan_ports(cur_addr)

  
```
