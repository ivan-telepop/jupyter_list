```typescript
Разница

1. Расширение

interface можно расширять через extends
type — через пересечение &
interface A { a: number }
interface B extends A { b: number }

type A = { a: number }
type B = A & { b: number }

2. Декларативное объединение (merge)

interface может автоматически объединяться:
interface A { a: number }
interface A { b: number } // объединится
type — нет (будет ошибка)

3. Возможности

type гибче:
union (A | B)
tuple
примитивы
type ID = string | number
interface — только объектные структуры

4. Когда что использовать

interface — для описания объектов, API, классов
type — когда нужны union, комбинации, сложные типы

Если у тебя просто DTO типа userResp — разницы почти нет.
```
