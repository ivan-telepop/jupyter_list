```typescript
// typical factorial
function factorial(n){
    if(n === 1){
        return 1;
}
return n * factorial(n-1);
}


factorial(5)
```

```typescript
function fibonacci(n) {
    if(n <=1){
        return n;
    }else{
    return fibonacci(n - 1) + fibonacci(n - 2);
    }
}

let myArr = [8,9,10,11];

// myArr.map(fibonacci)

for (let y of myArr){
    console.log(y,"Y")
}
```

```typescript
// factorial with cycle

function factIter(n) {
    let res = 1;
    for (let i = 2; i <=n; i++){
        res *=i;
    }
    return res;
}

factIter(6)
```
