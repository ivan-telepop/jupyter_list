```typescript
Arrays:
Ordered, indexed lists that can store multiple items of any data type.
Elements are accessed by numerical indices, starting from 0.
Example:
JavaScript

        const myArray = [1, 'hello', true];
        console.log(myArray[0]); // Output: 1
Maps:
Collections that store key-value pairs, similar to dictionaries in other languages.
Keys can be of any data type (unlike plain objects, which primarily use string keys).
Maintain the insertion order of elements.
Example:
JavaScript

        const myMap = new Map();
        myMap.set('name', 'Alice');
        myMap.set(1, 'one');
        console.log(myMap.get('name')); // Output: Alice
Sets:
Collections that store unique items, meaning no duplicate values are allowed. 
Useful when you need to ensure a collection contains only distinct elements. 
Example:
JavaScript

        const mySet = new Set([1, 2, 2, 3]);
        console.log(mySet); // Output: Set { 1, 2, 3 }
Additionally, JavaScript provides specialized collection types for specific use cases:
WeakMap: Similar to Map, but its keys are held "weakly," meaning if there are no other references to a key object, 
    it can be garbage collected. This prevents memory leaks in certain scenarios.
WeakSet: Similar to Set, but its values (which must be objects or symbols) are held "weakly," allowing for garbage collection if no 
    other references exist.
HTMLCollection / NodeList: These are array-like objects returned by DOM methods (e.g., getElementsByTagName, querySelectorAll). 
        They represent collections of HTML elements in the document.
```
