```typescript
// The first .then() receives the Response object. 
    // This runs even for HTTP error statuses (like 404 Not Found or 500 Internal Server Error).
    
    // Check if the response status is OK (status in the range 200-299)
          // If OK, parse the response body (e.g., as JSON or text). 
    // The .json() method also returns a promise, so we return it to chain the next .then().
      

let URL = 'http://localhost:8000/users'

fetch(URL).then(response => {if (!response.ok) {throw new Error('Network response was not ok: ' + response.statusText);}
                             return response.json();
  })
  .then(data => {
    // The second .then() receives the parsed data from the previous .then()'s return value.
    console.log('Fetched data:', data);
  })
  .catch(error => {
    // The .catch() block handles any errors that occurred in the entire chain 
    // (network errors during fetch() itself, or errors thrown manually).
    console.error('There was a problem with the fetch operation:', error);
  });
```
