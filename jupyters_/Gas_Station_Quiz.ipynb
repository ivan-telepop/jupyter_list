```python
# day items dictionary
day_items = dict(
    left_col=300.71,   # левая колонка литры
    right_col=301.65, # правая колонка литры
    low_price_liters=308.61, # литры по сниженной цене
    cashless_amount=3245 # сумма по безналичному расчету
)


def calculate_day(day_items):
    """Method takes as an argument:
    agrs: day_items: dict
    returns: float (calculated amount of money)
    """
    both_cols_amount = (day_items['left_col'] + day_items['right_col']) * 28 # All roubles
    minus_low = (day_items['low_price_liters'] * 2) # Roubles that sould be substracted from whole amount
    return both_cols_amount - day_items['cashless_amount']  - minus_low
        


calculate_day(day_items)   #490.41,235.03,288.23,5552)
```

```python

class DataItems:
    """Class that consist of values or fields depending on day report data"""
    def __init__(self,left_col,right_col,low_price_liters,cashless_amount):
        self.left_col: float = left_col
        self.right_col:float = right_col
        self.low_price_liters:float = low_price_liters
        self.cashless_amount:float = cashless_amount
        

class DayReport:
    """Class that makes report by calculating fields from DataItems class"""
    def __init__(self,date):
        if isinstance(date, DataItems):
            self.date = date
        else:
            raise ValueError("No DataItems object")
    def eval_report(self):
        both_cols_amount = (self.date.left_col + self.date.right_col) * 28 
        minus_low = (self.date.low_price_liters * 2) 
        return both_cols_amount - self.date.cashless_amount - minus_low
        
        
        


day = DataItems(490.41,235.03,288.23,5552)

rep = DayReport(day)





rep.eval_report()

```
