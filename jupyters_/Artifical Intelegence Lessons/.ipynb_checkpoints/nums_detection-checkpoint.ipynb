```python
import random

num0 = list('111101101101111')
num1 = list('001001001001001')
num2 = list('111001111100111')
num3 = list('111001111001111')
num4 = list('101101111001001')
num5 = list('111100111001111')
num6 = list('111100111101111')
num7 = list('111001001001001')
num8 = list('111101111101111')
num9 = list('111101111001111')

nums = [num0,num1,num2,num3,num4,num5,num6,num7,num8,num9]

# all nums

tema = 5 # Theme of lesson
n_sens = 15 # sensors quantity
weights = [ 0 for i in range(n_sens)] # INITIAL STATE OF WEIGHTS - 15 zeros


def perseptio(Sensor): # If got correct does True
    b = 7
    s = 0
    for y in range(n_sens):
        s += int(Sensor[y]) * weights[y]
        if s >= b:
            return True
        else:
            return False

# decreases weights
def decrease(number):
    for i in range(n_sens):
        if int(number[i]) == 1:
            weights[i] -=1

# increases weights
def increase(number):
    for i in range(n_sens):
        if int(number[i]) == 1:
            weights[i] +=1


# repeating cycles
n = 1000

for i in range(n):
    j = random.randint(0,9) # generating random
    r = perseptio(nums[j]) # True

    if j != tema:
        if r:
            decrease(nums[j])
    else:
        if not r:
            increase(nums[tema])



# Atempts to make detection

# print(f"0 - is : {tema} correct ?", perseptio(num0))
# print(f"1 - is : {tema} correct ?", perseptio(num1))
# print(f"2 - is : {tema} correct ?", perseptio(num2))
# print(f"3 - is : {tema} correct ?", perseptio(num3))
# print(f"4 - is : {tema} correct ?", perseptio(num4))
# print(f"5 - is : {tema} correct ?", perseptio(num5))
# print(f"6 - is : {tema} correct ?", perseptio(num6))
# print(f"7 - is : {tema} correct ?", perseptio(num7))
# print(f"8 - is : {tema} correct ?", perseptio(num8))
# print(f"9 - is : {tema} correct ?", perseptio(num9))

for one in range(len(nums)):
    print(f"Trying {one} : {nums[one]} - is:",perseptio(nums[one]))
```
