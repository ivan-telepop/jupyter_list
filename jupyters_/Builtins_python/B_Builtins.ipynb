```python
# int to binary representation 
int_lst = [ bin(i) for i in range(1,20)]


# back again - (base param 2)
for y in int_lst:
    print(int(y,2))


res = zip(int_lst, [ int(x,2) for x in int_lst ])

print(*res)



# for example we have an ip address
ip = [5,231,77,19]

bip = zip([i for i in ip], [ bin(i) for i in ip])


print(*bip)
```

```python
# Функция останова breakpoint() вызывает функцию sys.breakpointhook(). По умолчанию sys.breakpointhook() в 
# свою очередь вызывает функцию pdb.set_trace().

# Использование встроенной функции breakpoint(), по крайней мере, обеспечивает удобство использования отладчика, 
# поскольку не нужно явно импортировать модуль pdb, а так же писать дополнительный код, чтобы войти в отладчик. 
# Функция breakpoint() представляет синтаксический сахар над имеющимися конструкциями


from pprint import pprint
import sys

def print_locals():
    caller = sys._getframe(1)  
    pprint(caller.f_locals)

x = 10
y = 'Hi'
z = 'Hello'
breakpoint()
```
