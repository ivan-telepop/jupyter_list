```python
import struct
def float_to_IEEE_754(f : float) -> str:
    s=''.join(f'{b:08b}' for b in struct.pack('d',f)[::-1])
    return f"{s[0]} {s[1:12]} {s[12:]}"

float_to_IEEE_754(0.245)
```

```python
import string
# string.ascii_letters
# Output: abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ
# -----
# string.ascii_lowercase
# Output: abcdefghijklmnopqrstuvwxyz
# -----
#string.ascii_uppercase)
# Output: ABCDEFGHIJKLMNOPQRSTUVWXYZ
# ----
# string.digits
# Output: 0123456789

letters_order = [ord(i) for i in string.ascii_letters]

numerical_order = [ord(n) for n in string.digits]

print("Letters:",leters_order)
print("Digits:", numerical_order)

bytes_nums = [ bin(i)[2:] for i in numerical_order]
bytes_letters = [ bin(i)[2:] for i in letters_order]

print("Nums as Bytes:", bytes_nums)
print("ASCII Letters as Bytes:", bytes_letters)

```

```python
# Основные побитовые операторы
# << (Побитовый сдвиг влево): Сдвигает биты числа 
#  влево на 
#  позиций. Эквивалентно умножению на 
# .
# >> (Побитовый сдвиг вправо): Сдвигает биты вправо на 
#  позиций. Эквивалентно целочисленному делению на 
# .
# & (И, AND): Устанавливает бит в 1, если он равен 1 в обоих операндах.
# | (ИЛИ, OR): Устанавливает бит в 1, если он равен 1 хотя бы в одном из операндов.
# ^ (Исключающее ИЛИ, XOR): Устанавливает бит в 1, если он равен 1 только в одном из операндов, но не в обоих.
# ~ (НЕ, NOT): Инвертирует все биты числа (0 становится 1, 1 становится 0). Результат 
#  равен 

```

```python
import base64

base64_string = "SGVsbG8gV29ybGQ="

# Decode from Base64 bytes and turn back into readable text
original_string = base64.b64decode(base64_string).decode('utf-8')

print(original_string)
```
