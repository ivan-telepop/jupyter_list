```python
# Recursive Cases and concept
# Think about for gettting used this concept so operate in mind 

def multy(n1,n2):
    if n1 == 0 or n2 == 0: # Base case
        return 0
    return n1 + multy(n1,n2 -1)


multy(6,2)
```

```python
# Алгоритм Евклида 
# Нахождение НОД для двух целых чисел (наибольший общий делитель)
# Представим что мы имеем прямоугольник размером А х В 
# его нужно поделить на квадраты, но квадраты должны быть максимально большими.
# Функция отвечает на вопрос - какой размер квадрата (общего делителя).
# Вариант с рекурсией.


def euclidus(a,b):
    r = a % b 
    if r == 0:
        return f"The biggest square are :{b} x {b}"
    else:
        a,b = b,r
        return euclidus(a,b)

euclidus(1800,600)
```

```python
# Recursive generator
def recursive_gen(n):
    if n == 0:
        yield 0
    else:
        yield n
        yield from recursive_gen(n - 1)

# Используем генератор
gen = recursive_gen(5)

# for value in gen:
#     print(value)

print(next(gen))
print(next(gen))
print(next(gen))
```

```python
nested = [123,22,33,44,55,[77,888,878,44,333,[333,55,77,[75757,2442,959,[5353,7373,42424]]]]]


def flatten(nested_list):
    result = []
    for item in nested_list:
        if isinstance(item, list): # We do check 
            result.extend(flatten(item))  # Recursive call for nested lists
        else:
            result.append(item)
    return result


foo = flatten(nested)

print(foo)


```

```python
k = [ i for i in 'ASDFGHJKLQWERTY']
v = [ ord(x) for x in k]

n_dict = dict(zip(k,v))

n_dict['G'] = dict(zip(list('ZXCVB'), list(range(1,len('ZXCVB')))))

n_dict['X'] = dict(zip(list('LKJHGFD'), list(range(1,len('LKJHGFD')))))


def nested_clean(d):
    cleaned = {}
    for k,v in d.items():
        if isinstance(v,dict):
            cleaned.update(nested_clean(v))
        else:
            cleaned.update(d.items())
    return cleaned


nested_clean(n_dict)
```
