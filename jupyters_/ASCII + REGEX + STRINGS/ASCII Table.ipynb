```python
a = b'1234567890' # From sequnce - a

b = b'0987654321' # Symbols Will be changed to the symbls from sequnce - b

table = bytes.maketrans(a,b)

wrd = b'1234567890'

wrd.translate(table)
```

```python
foo = 'qwerty'

A = bytes('qwertyu','utf-8')
B = bytes('QWERTY{','utf-8')


# Тут как бы схема или таблица переобразований - из A символы меняем на символы которые в B
tran = bytes.maketrans(A,B)

# А тут обьект с которым работаем 
'q-w-e-r-t-y-u'.translate(tran)
```
